```python
import numpy as np
import pandas as pd

# 1. 模拟数据集
data = {
    "User_ID": [1, 2, 3, 4, 5, 6],
    "Age": [23, 45, np.nan, 31, np.nan, 38],
    "Income": [50000, 62000, 58000, 999999, 52000, 65000],
}

df = pd.DataFrame(data)

# --- 你的代码开始 ---
df["Age"] = df["Age"].fillna(int(df["Age"].mean()), inplace=False)

df.loc[df["Income"] > 500000, "Income"] = df["Income"].median()


# 提示：
# - 填充缺失值：df['列名'].fillna(值, inplace=True) 或者 df['列名'] = df['列名'].fillna(值)
# - 计算平均值和中位数：df['列名'].mean(), df['列名'].median()
# - 条件定位：df.loc[df['列名'] > 阈值, '列名'] = 新值


# --- 你的代码结束 ---
df
```
